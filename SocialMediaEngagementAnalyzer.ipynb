```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = {
    'Post_ID': [1,2,3,4,5],
    'Platform': ['Instagram','Facebook','Twitter','Instagram','Facebook'],
    'Likes': [120, 80, 45, 200, 150],
    'Comments': [30, 10, 5, 50, 20],
    'Shares': [15, 5, 2, 25, 10]
}

df = pd.DataFrame(data)
df
```

```python
df['Engagement_Score'] = df['Likes'] + df['Comments'] + df['Shares']
df
```

```python
platform_engagement = df.groupby('Platform')['Engagement_Score'].mean()
platform_engagement
```

```python
top_post = df.loc[df['Engagement_Score'].idxmax()]
top_post 
```

```python
platform_engagement.plot(kind='bar')
plt.title("Average Engagement by Platform")
plt.xlabel("Platform")
plt.ylabel("Engagement Score")
plt.show()
```
